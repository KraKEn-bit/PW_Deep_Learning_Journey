```python
from tensorflow import keras
from keras.datasets import mnist
from tensorflow.keras.layers import Conv2D, MaxPooling2D, AveragePooling2D
from keras.layers import Dense, Flatten
from keras.models import Sequential
```

```python
(x_train, y_train), (x_test, y_test) = mnist.load_data()
# Peforming reshaping operation
x_train = x_train.reshape(x_train.shape[0], 28, 28, 1)
x_test = x_test.reshape(x_test.shape[0], 28, 28, 1)
```

```python
x_train.shape[0]
```

```python
x_test.shape[0]
```

#**Normalization:**

```python
x_train = x_train / 255
x_test = x_test / 255
```

#**One Hot Encoding of the Labels:**

```python
y_train = keras.utils.to_categorical(y_train, 10)
y_test = keras.utils.to_categorical(y_test, 10)
```

```python
y_test[100]
```

#**Network Structure:**

```python
# LeNet-5 Model Architecture:

model = Sequential()

model.add(Conv2D(6, kernel_size=(5, 5), activation='relu', input_shape=(28, 28, 1))) # output: 24 x 24 x 6
model.add(MaxPooling2D(pool_size=(2, 2))) # Using the previous o/p through this layer I get = 12 x 12 x 6
model.add(Conv2D(16, kernel_size=(5, 5), activation='relu')) # Similarlt I get O/p: 8 x 8 x 16
model.add(MaxPooling2D(pool_size=(2, 2))) # Then O/p = 4 x 4 x 16
model.add(Conv2D(120, kernel_size=(5, 5), activation='relu', padding='same')) # O/p = 4 x 4 x 120
model.add(Flatten())
model.add(Dense(84, activation='relu'))
model.add(Dense(10, activation='softmax'))
```

```python
model.summary()
```

```python
model.compile(loss=keras.metrics.categorical_crossentropy, optimizer=keras.optimizers.Adam(), metrics=['accuracy'])
# Why use Categorical_Crossentropy? - Because we use MultiClass Classification

model.fit(x_train, y_train, batch_size=128, epochs=20, verbose=1, validation_data=(x_test, y_test))
```

#**My RESULTS:**

```python
score = model.evaluate(x_test, y_test)
print('Test Loss:', score[0])
print('Test accuracy:', score[1])
```
