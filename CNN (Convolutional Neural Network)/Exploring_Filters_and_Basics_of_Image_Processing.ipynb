```python
# Reading the image:
image = mpimg.imread('input.jpg')
plt.imshow(image)
```

```python
# Isolating RGB channels
r = image[:,:,0]
g = image[:,:,1]
b = image[:,:,2]



# Visualizing the individual color channels
f, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(20,10))
ax1.set_title('R channel')
ax1.imshow(r, cmap='gray')
ax2.set_title('G channel')
ax2.imshow(g, cmap='gray')
ax3.set_title('B channel')
ax3.imshow(b, cmap='gray')
```

# **Converting to grayscale for filtering:**

```python
gray = cv2.cvtColor(image, cv2.COLOR_RGB2GRAY)

plt.imshow(gray, cmap='gray')
```

###**If I remove the "cmap='gray'"**:

```python
gray = cv2.cvtColor(image, cv2.COLOR_RGB2GRAY)

plt.imshow(gray)
```

# **Using Filters:**

```python
import numpy as np

#Horizontal Edge Detection
sobel_y = np.array([[ -1, -2, -1],
                   [ 0, 0, 0],
                   [ 1, 2, 1]]) # 1st Filter


# vertical edge detection
sobel_x = np.array([[-1, 0, 1],
                   [-2, 0, 2],
                   [-1, 0, 1]]) # 2nd Filter


# filtering the image using filter2D(grayscale image, bit-depth, kernel)
filtered_image1 = cv2.filter2D(gray, -1, sobel_x)
filtered_image2 = cv2.filter2D(gray, -1, sobel_y)


f, ax = plt.subplots(1, 2, figsize=(15, 4))
ax[0].set_title('vertical edge detection')
ax[0].imshow(filtered_image1, cmap='gray')
ax[1].set_title('horizontal edge detection')
ax[1].imshow(filtered_image2, cmap='gray')
```

```python
# filtering the image using filter2D(grayscale image, bit-depth, kernel)
filtered_image1 = cv2.filter2D(gray, -1, sobel_x)
filtered_image2 = cv2.filter2D(gray, -1, sobel_y)


f, ax = plt.subplots(1, 2, figsize=(15, 4))
ax[0].set_title('vertical edge detection')
ax[0].imshow(filtered_image1)
ax[1].set_title('horizontal edge detection')
ax[1].imshow(filtered_image2)
```
