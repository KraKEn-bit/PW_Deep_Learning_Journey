#**Tensor Operation and Gradients:**

```python
x = torch.tensor(3.) # Input
w = torch.tensor(4. , requires_grad = True) # Weight
b = torch.tensor(5., requires_grad = True) # Bias

x, w, b
```

#**Arithmatic Operation:**

```python
y = w*x + b
y
```

#**Compute Derivates:**

```python
y.backward()
```

#**Displaying Gradients:**

```python
print("dy/dx: ",x.grad)
print("dy/dw: ",w.grad)
print("dy/db: ",b.grad)
```

# **Interoperability with Numpy:**

```python
import numpy as np

x = np.array([[1,2], [3, 4.]])
x
```

### **Numpy array into a PyTorch Tensor**:

```python
y = torch.from_numpy(x) # converts a NumPy array into a PyTorch tensor.
y
```

```python
x.dtype, y.dtype
```

###**PyTorch to Numpy:**

```python
z = y.numpy()
z
```
